### Why use git?

**Answer**: Sustainable Software Development

### What is Sustainable Software Development?

Generally, when we say sustainable software, we mean software that: 

1. Is useful for more than a single task (like having a publication quality plotting program you can use for all plots instead of redoing it every time),
1. Is shareable (so other people can read and understand it, and perhaps use it), 
1. It works not just on your computer with your particular settings (for file locations, Python libraries, etc.) 
1. Finally, if the software is used by a larger community, then that community should know how to interact with the code and each other.

Android, R, Python, SciPy, and matplotlib are all examples of sustainable software development!

## A BRIEF INTERLUDE TO GET GIT SETUP

### Github Account

Go to [github.com](http://github.com) and make an account (if you haven't already done so).

As a quick, but important, aside: *git* is the name of the software system that we're using. *Github* is a hosting service that we're using to store our remote repos. There are other hosting services&mdash;like Gitlab and Bitbucket&mdash;but we'll be sticking with Github for this class.

### <font color="red">IMPORTANT:</font> Generating a Personal Access Token for GitHub

In order to clone private repositories and push/pull changes to those repositories on GitHub, you will also need a Personal Access Token (PAT) to access GitHub via the terminal (on your computer or elsewhere). **When typing the PAT as the password it will not appear on the command line, but it is being entered. This is a security feature so that your password is not visible**

&#9989; **Task 11**

<font size=+3>&#128064;</font> Get your own PAT by following the instructions from  [**this video.**](https://mediaspace.msu.edu/media/Setting+up+a+personal+access+token+on+Github.com/1_exthwtyb). The menus have changed a bit, but the flow is the same.

**Useful suggestions**: Once you've set up a PAT and you've confirmed that you can use it to access GitHub, we have a couple of suggestions:

1. Save it somewhere safe (like in a password manager or in a file/note on a secure computer) because you'll need it again later.
2. You may wish to store your GitHub login credentials in your git configuration file so that you don't have to enter them every time you want to push/pull changes to/from GitHub. To do this, you can run the following command in your terminal:
    
    ``` bash
    git config --global credential.helper store
    ```
    **Note**: If working locally and you're using a Mac, you may need to use the following command instead of the one above:
    ``` bash
    git config --global credential.helper osxkeychain
    ```
**Note 1:** At first it will *appear* to do nothing, but it will be ready to remember your PAT once you enter it the first time.
