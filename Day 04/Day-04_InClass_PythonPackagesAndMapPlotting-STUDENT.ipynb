````python
import folium
import pandas as pd

mo_geo = 'MO_County_Boundaries.json'
mo_cancerfile = 'cancer18to22_incidence_ageadj_per100k.csv'

mo_data = pd.read_csv(mo_cancerfile)

#Let Folium determine the scale

mo_map = folium.Map([38.5, -92.5], zoom_start=7)
folium.Choropleth(geo_data=mo_geo, data=mo_data,
             columns=['County', 'Incidence'],
             key_on='feature.properties.COUNTYNAME',
             fill_color='YlOrRd', fill_opacity=0.7, line_opacity=0.2,
             legend_name='Unemployment Rate (%)', smooth_factor=0.1,highlight=True).add_to(mo_map)
folium.LayerControl().add_to(mo_map)

mo_map
````
