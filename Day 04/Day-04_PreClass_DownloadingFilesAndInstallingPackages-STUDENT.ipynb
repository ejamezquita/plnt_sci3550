### 2.3 A word on Python environments and `conda` versus `pip`

#### Always use virtual environments

Regardless of what installation method you choose, it is good practice to **never** install packages in your `(base)` Python. Always create an virtual environment and install whatever you need in there, instead. If you installed Python following the class setup guides, that's what we did with the command:

```bash
conda create -n myenv anaconda::jupyter ipython matplotlib numpy pandas ...
```

There, we created an environment named `myenv`. And while we were at it, we also installed a bunch of useful packages like NumPy and matplotlib. Two main advantages of doing virtual environments: 

- You want to keep two versions of the same package. Sometimes, newer package updates will *remove* functions that we need. So we can keep a dedicated environment with old software.
- Two or more packages are incompatible with each other and everything is screaming errors. We can simply delete the environment&mdash;not Python&mdash;and reinstall all the packages one by one to detect where things went awry.

You can read more details on [virtual environments here](https://www.anaconda.com/docs/getting-started/working-with-conda/environments).

#### `conda` versus `pip`

In general, if you are using a conda-based Python, you install stuff with `conda` as much as possible. As you noticed before, both `conda` and `pip` will also update other packages when needed. But `conda` does **not** keep track of stuff modified by `pip`.

> Because of `conda`’s lack of awareness of environment updates made by `pip`, using `pip` in your environment must be the last action that will be performed when building the environment. [More details here](https://www.anaconda.com/docs/getting-started/working-with-conda/packages/pip-install).

One more reason to stick to virtual environments: by mixing `conda` and `pip` installs you can end up with a nasty mess. A virtual environment is a way to contain such mess.

#### Going back to older versions of packages

In rare, though not impossible, circumstances, you may need to install an *older* version of a package. This can happen when a package requires an older version of another package. (Again, rare, but not crazy rare). In this case, we still use `conda`, but now we specify the version, like so:

```bash
conda install matplotlib=3.9 #with conda, use a single =

pip install matplotlib==3.9 #with pip, use double ==
```

#### Checking for outdated packages

Now that you've practiced updating a package, you can also check to see if any of your packages are out of date. To do this, you can use the following command:

```bash
conda search --outdated # with conda

pip list --outdated # with pip
```

This can be useful if you want to make sure you're always using the most up to date version of a package. Or, if you're working on a project with a collaborator, you can make sure you're both using the same version of a package, which can help avoid some headaches.
